## 2: Check the size and structure

## 3: Check data types

Each column in a dataset has a type, such as text, whole number, or decimal number. 
Here we check what type pandas has assigned to each column. 
This helps us catch columns that look numeric but are stored as text, which is a common issue.

```python
df.info()
```

### Note on no_ratings and no_reviews

Both no_ratings and no_reviews are stored as text instead of numbers. 
At first glance most values look like plain numbers, for example 900. 
However, a small number of rows contain commas as thousands separators, for example 15,210. 
Because pandas requires every value in a column to fit one single type, 
the presence of even a few comma separated values forces the entire column to be stored as text. 
Both columns will need to be cleaned by removing commas and converting to numeric type 
before they can be used in any calculations.

```python
print(df["no_ratings"].str.contains(",").sum(), "values in no_ratings contain a comma")
print(df["no_reviews"].str.contains(",").sum(), "values in no_reviews contain a comma")
```

## 4: Check for missing values

## 8: Investigate duplicate rows before deciding what to do

Before removing duplicate rows, we look at a sample of them. 
This helps us understand whether they are truly repeated entries or just short, generic reviews written by different people.

```python
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows.sort_values("review").head(20)
```

### Note on duplicate reviews

Investigation of the duplicate rows shows that most duplicates are long, detailed, uniquely worded reviews 
that appear multiple times under the same product, with identical text, rating, and formatting. 
It is highly unlikely that different customers independently wrote identical multi-sentence reviews. 
This suggests the duplication happened during data collection, most likely the scraper captured the 
same review more than once. 
Since these are not distinct opinions from different people, it is appropriate to remove exact 
duplicate rows during the cleaning stage using drop_duplicates.
